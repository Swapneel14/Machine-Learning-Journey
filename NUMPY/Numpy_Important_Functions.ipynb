#Set Functions:-

```python
x = np.array([1,2,3,4,56,67])
y = np.array([2,3,4,5,6,7,8])

print(np.union1d(x,y))
print(np.intersect1d(x,y))
print(np.setdiff1d(x,y))
print(np.setxor1d(x,y))
```
