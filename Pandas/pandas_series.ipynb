```python
import numpy as np
import pandas as pd
```

#Pandas Series:-

```python
country = ["India","America","Japan","Belgium"]
pd.Series(country)
```

```python
#custom index
marks = [99,100,23]
subs=["Maths","DSA","Algo"]

pd.Series(marks,index=subs)
```

```python
#setting a name
pd.Series(marks,index=subs,name="Mere Marks")
```

#Series from Dict

```python
marks={
    "Maths":99,
    "Geo":94,
    "Physics":100
}

marks_series=pd.Series(marks,name="My Marks")
print(marks_series)
```
